# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

Word2Vec consists of models for generating word embedding. These models are two-layer neural networks having one input layer, one hidden layer, and one output layer.

Word2Vec utilizes two architectures :

````python
print("Cosine similarity between 'provide' and 'program' - Skip Gram : ", w2v.wv.similarity('provide', 'program'))
````

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd

docs = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?'
]

vec = TfidfVectorizer()
X = vec.fit_transform(docs)

sim = cosine_similarity(X)
names = ['doc1', 'doc2', 'doc3', 'doc4']
print(pd.DataFrame(sim, index=names, columns=names).round(3))

# doc1 and doc4 get a similarity of 1.0. They have exactly the same words, only the order is different, and TF-IDF ignores order. doc3 is the one that is least like the others.
````

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

sentences = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data'
]

tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(sentences)

tfidf_df = pd.DataFrame(matrix.toarray(), columns=tfidf.get_feature_names_out())
print(tfidf_df.round(3).T)

# the three highest-weighted words of every sentence
for i, row in tfidf_df.iterrows():
    print(f"sentence {i + 1}:", row.sort_values(ascending=False).head(3).round(3).to_dict())

# The important words are the ones that appear in one sentence but not in the others: *science* and *fields* in the first sentence, *best* and *courses* in the second, *scientists* and *analyze* in the third. *data* shows up everywhere so its weight is lower, except in the third sentence where it is repeated twice.
````

````python
import re
import contextlib
import io
import sys
import pandas as pd
import gensim

simpsons = pd.read_csv('simpsons_script_lines.csv', on_bad_lines='skip', low_memory=False)
simpsons = simpsons.dropna(subset=['spoken_words'])     # some lines have no spoken words
print(simpsons.shape)

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

simpsons['clean'] = simpsons['spoken_words'].apply(clean_text)
tokens = [line.split() for line in simpsons['clean']]
tokens[:3]

_word2vec_stderr = io.StringIO()
with contextlib.redirect_stderr(_word2vec_stderr):
    Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1, workers=4)

_known_dot_message = "Exception ignored in: 'gensim.models.word2vec_inner.our_dot_float'\n"
_remaining_stderr = _word2vec_stderr.getvalue().replace(_known_dot_message, "")
if _remaining_stderr:
    sys.stderr.write(_remaining_stderr)
print("words in the vocabulary:", len(Skip_gram_model.wv))
````

### Task 4

Use: wv.most_similar() method to :

````python
for name in ['homer', 'marge', 'bart']:
    print(f"words similar to '{name}':")
    for word, score in Skip_gram_model.wv.most_similar(name):
        print(f"   {word:15} {score:.3f}")
    print()
````
